Примечание: Ввод и вывод уже написан за вас.

```python
# Формат ввода:

# В первой строке идет число n <= 100 — количество моделей, которые участвуют в блендинге. 
# Во второй строке через пробел записаны n чисел — значения точности каждой из этих моделей на лидерборде (от 0 до 1).
# Формат вывода:

# В единственной строке выведите n чисел — веса, с которыми нужно взвесить соответствующие модели. 
# Значения весов округлите до 7 знака после запятой. Исходный порядок моделей важен. 

import numpy as np

n = 3
s = '0.3 0.5 0.4'.split()
scores = []
for i in range(n):
    scores.append(float(s[i]))

## Your code here ...
weights = np.round((np.array(scores) / np.sum(scores)), 7)
weights
```

**🤝Soft Voting**

Датасета с предсказаниями поломок машин от разных моделей:

```python
cb_preds = pd.read_csv('../data/catboost_proba.csv')
cb_preds.head()
```

```python
lgbm_preds  = pd.read_csv('../data/lgbm_proba.csv')
lgbm_preds .head()
```

```python
xgb_preds = pd.read_csv('../data/xgb_proba.csv')
xgb_preds.head()
```

```python
final_preds = sum([cb_preds, lgbm_preds , xgb_preds])
final_preds.head()
```

```python
final_preds['soft_preds'] = final_preds.idxmax(axis=1)
result = final_preds[['soft_preds']].map(lambda x: x.split('_proba')[0])
result
```
