В ноутбуке этого урока мы реализовали функцию reduce_mem_usage

```python
def reduce_mem_usage(df):
    """ iterate through all the columns of a dataframe and modify the data type
        to reduce memory usage.
    """
    start_mem = df.memory_usage().sum() / 1024**2
    print('Memory usage of dataframe is {:.2f} MB'.format(start_mem))

    for col in df.columns:
        col_type = df[col].dtype.name

        if col_type not in ['object', 'category', 'datetime64[ns, UTC]']:
            c_min = df[col].min()
            c_max = df[col].max()
            if str(col_type)[:3] == 'int':
                if c_min > np.iinfo(np.int8).min and c_max < np.iinfo(np.int8).max:
                    df[col] = df[col].astype(np.int8)
                elif c_min > np.iinfo(np.int16).min and c_max < np.iinfo(np.int16).max:
                    df[col] = df[col].astype(np.int16)
                elif c_min > np.iinfo(np.int32).min and c_max < np.iinfo(np.int32).max:
                    df[col] = df[col].astype(np.int32)
                elif c_min > np.iinfo(np.int64).min and c_max < np.iinfo(np.int64).max:
                    df[col] = df[col].astype(np.int64)
            else:
                if c_min > np.finfo(np.float16).min and c_max < np.finfo(np.float16).max:
                    df[col] = df[col].astype(np.float16)
                elif c_min > np.finfo(np.float32).min and c_max < np.finfo(np.float32).max:
                    df[col] = df[col].astype(np.float32)
                else:
                    df[col] = df[col].astype(np.float64)

    end_mem = df.memory_usage().sum() / 1024**2
    print('Memory usage after optimization is: {:.2f} MB'.format(end_mem))
    print('Decreased by {:.1f}%'.format(100 * (start_mem - end_mem) / start_mem))

    return df
```

**Генераторы**

Напишите генератор, который принимает массив натуральных чисел и максимально возможное число в нем, а потом на каждой итерации возвращает OHE следующего элемента в массиве. Возвращаемый массив должен быть размерности (max_class + 1,). <br>
Примечание: OHE (one hot encoding) - это способ преобразования одного числа в вектор (часто используется для кодировки классов). <br>
Во время него, вы числу x сопоставляете массив длины max_len, в которой на позиции x стоит 1, а на остальных позициях 0.

```python
import numpy as np

def ohe_generator(numbers, max_class):
    for num in numbers:
        # Создаём нулевой вектор длины max_class + 1
        ohe_vector = np.zeros(max_class + 1, dtype=int)
        # Устанавливаем 1 на позиции num
        ohe_vector[num] = 1

        yield ohe_vector

gen = ohe_generator([1, 3, 4], 5)

print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# У вас есть датафрейм df_cars, в котором есть два столбца: int_model и target_reg. 



# Задача: Написать функцию, которая для каждой группы (по int_model) 
# будет считать среднее, сумму, минимум и кол-во элементов в группе. 
# Для подсчета данных значений не используйте pd.groupby(), а используйте векторизацию через numpy.

# Функция должна возвращать датафрейм в следующем формате:

# Подсказка 1
# Кол-во, сумму и среднее можно посчитать при помощи np.bincount.

# Подсказка 2
# Минимум можно посчитать при помощи np.minimum.reduceat.
```

```python
import numpy as np
import pandas as pd

def fast_groupby(df_cars):
    # Извлекаем значения столбцов
    int_model = df_cars['int_model'].values
    target_reg = df_cars['target_reg'].values

    # Получаем уникальные группы и индексы групп для каждой строки
    unique_groups, group_idx = np.unique(int_model, return_inverse=True)
    n_groups = len(unique_groups)

    # 1. Количество элементов в каждой группе
    count_values = np.bincount(group_idx)

    # 2. Сумма target_reg по группам
    sum_values = np.bincount(group_idx, weights=target_reg)

    # 3. Среднее (сумма / количество)
    mean_values = sum_values / count_values

    # 4. Минимум по группам — через reduceat
    # Сортируем индексы по группам
    sorted_order = np.argsort(int_model)
    sorted_target = target_reg[sorted_order]
    sorted_group_idx = group_idx[sorted_order]

    # Находим границы групп
    split_indices = np.r_[0, np.where(np.diff(sorted_group_idx))[0] + 1]

    # Применяем minimum.reduceat для поиска минимума в каждой группе
    min_values = np.minimum.reduceat(sorted_target, split_indices)

    # Формируем результат
    res = pd.DataFrame({
        'int_model': unique_groups,
        'min': min_values,
        'sum': sum_values,
        'count': count_values,
        'mean': mean_values
    })

    return res
```
